# DBU Usage for My Specific Pipelines & Jobs

```python
%sql
--  DBU Usage for My Specific Pipelines & Jobs
SELECT
    usage_date,
    sku_name,
    -- Mapping specific SKU types to your resources
    CASE 
        WHEN sku_name LIKE '%DLT%' THEN 'DLT Pipelines (Gold/Silver/Chunk4)'
        WHEN sku_name LIKE '%JOBS%' THEN 'Workflows (Chunking/Copy/End-to-End)'
        ELSE 'Other'
    END AS resource_group,
    SUM(usage_quantity) AS total_dbus
FROM
    system.billing.usage
WHERE
    usage_date >= CURRENT_DATE() - INTERVAL 7 DAYS
    AND usage_unit = 'DBU'
GROUP BY 1, 2, 3
ORDER BY 1 DESC;
```

# Automated vs. Interactive (Manual) Cost Analysis

```python
%sql
--  Automated vs. Interactive (Manual) Cost Analysis
SELECT
    CASE 
        WHEN sku_name LIKE '%ALL_PURPOSE%' THEN 'Manual / Interactive (Expensive)'
        ELSE 'Automated Production (Optimized)'
    END AS run_type,
    SUM(usage_quantity) AS total_dbus,
    (SUM(usage_quantity) / SUM(SUM(usage_quantity)) OVER()) * 100 as cost_percentage
FROM
    system.billing.usage
WHERE
    usage_date >= CURRENT_DATE() - INTERVAL 30 DAYS
GROUP BY 1;
```

# DBU Consumption Trend by Resource Type 

```python
%sql
--  DBU Consumption Trend by Resource Type 
SELECT 
    usage_date,
    -- Grouping into 3 clear categories
    CASE 
        WHEN sku_name LIKE '%DLT%' THEN 'DLT Pipelines'
        WHEN sku_name LIKE '%JOBS%' THEN 'Automated Jobs'
        ELSE 'Interactive/Manual'
    END AS usage_source,
    SUM(usage_quantity) as dbus
FROM 
    system.billing.usage
WHERE 
    usage_date >= CURRENT_DATE() - INTERVAL 14 DAYS
GROUP BY 1, 2
ORDER BY 1;
```

# Peak Usage Hours

```python
%sql
--  Peak Usage Hours 
SELECT 
    hour(usage_start_time) as usage_hour,
    dayofweek(usage_date) as day_of_week,
    SUM(usage_quantity) as dbus
FROM 
    system.billing.usage
WHERE 
    usage_date >= CURRENT_DATE() - INTERVAL 7 DAYS
GROUP BY 1, 2
ORDER BY 3 DESC;
```
